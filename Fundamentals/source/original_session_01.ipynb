```python
import numpy as np
import matplotlib.pyplot as plt

def wave(x, kind="sin"):
    if kind == "sin":
        return np.sin(x)
    elif kind == "cos":
        return np.cos(x)

x = np.linspace(0, 2 * np.pi, 200)

y1 = wave(x, "sin")
y2 = wave(x, "cos")

plt.plot(x, y1, label="sin")
plt.plot(x, y2, label="cos")

plt.legend()
plt.grid(True)
plt.show()
```

```python
while True:
    print("\n--- ماشین حساب ساده ---")
    print("1. جمع")
    print("2. تفریق")
    print("3. ضرب")
    print("4. تقسیم")
    print("0. خروج")

    choice = input("انتخاب کن: ")

    # خروج از برنامه
    if choice == "0":
        print("خروج از ماشین حساب 👋")
        break

    # بررسی انتخاب درست
    if choice not in ["1", "2", "3", "4"]:
        print("گزینه اشتباهه! دوباره امتحان کن.")
        continue

    # گرفتن اعداد
    num1 = float(input("عدد اول را وارد کن: "))
    num2 = float(input("عدد دوم را وارد کن: "))

    # انجام محاسبات
    if choice == "1":
        result = num1 + num2
        print("نتیجه:", result)

    elif choice == "2":
        result = num1 - num2
        print("نتیجه:", result)

    elif choice == "3":
        result = num1 * num2
        print("نتیجه:", result)

    elif choice == "4":
        if num2 == 0:
            print("تقسیم بر صفر امکان‌پذیر نیست!")
        else:
            result = num1 / num2
            print("نتیجه:", result)
```
