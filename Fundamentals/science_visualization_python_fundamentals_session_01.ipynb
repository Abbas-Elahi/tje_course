<div class="sv-card teacher"><b>🧠 Mutable vs Immutable</b><br><b>Lists</b> are mutable. <b>Tuples and strings</b> are immutable. This difference matters when you update data.</div>

```python
a = [1, 2]
b = a

b.append(3)
print("a =", a)
print("b =", b)
```

```python
a = [1, 2]
b = a.copy()

b.append(3)
print("a =", a)
print("b =", b)
```

```python
a = [1, 2]
b = a
print(id(a) == id(b))
```

```python
import copy

matrix = [[1, 2], [3, 4]]
a = matrix.copy()
a[0][0] = 99

print("Original after shallow copy change:", matrix)
print("Shallow copy:", a)
```

```python
import copy

matrix = [[1, 2], [3, 4]]
b = copy.deepcopy(matrix)
b[0][0] = 99

print("Original after deep copy change:", matrix)
print("Deep copy:", b)
```

<div class="sv-card warning"><b>⚠️ Common Surprise</b><br>A shallow copy duplicates the outer list, but nested objects can still be shared. <code>deepcopy</code> recursively copies nested objects.</div>

<div class="sv-section"><div class="sv-section-title">🧩 6. Pattern Matching with match</div><div class="sv-section-sub">A clean alternative when one value can match several named cases.</div></div>

<div class="sv-card warning"><b>📌 Version Note</b><br><code>match / case</code> requires Python 3.10 or newer.</div>

```python
command = "start"
match command:
    case "start":
        print("Start game")
    case "exit":
        print("Exit")
```

```python
level = 5
match level:
    case 1:
        print("Easy")
    case 2:
        print("Medium")
    case _:
        print("Unknown")
```

```python
day = "Friday"
match day:
    case "Friday" | "Thursday":
        print("Holiday")
    case _:
        print("Working day")
```

```python
score = 18
match score:
    case x if x >= 17:
        print("Excellent")
    case x if x >= 10:
        print("Passed")
    case _:
        print("Failed")
```

<div class="sv-section"><div class="sv-section-title">🌊 11. Scientific Functions with NumPy + Matplotlib</div><div class="sv-section-sub">This is where formulas start to become pictures.</div></div>

<div class="sv-center"><img src="assets/science_pipeline.png" width="900"></div>

<div class="sv-center"><img src="assets/wave_preview.png" width="860"></div>

```python
import numpy as np
import matplotlib.pyplot as plt

def wave(x, kind="sin"):
    if kind == "sin":
        return np.sin(x)
    elif kind == "cos":
        return np.cos(x)
    else:
        raise ValueError("kind must be 'sin' or 'cos'")

x = np.linspace(0, 2 * np.pi, 200)
y1 = wave(x, "sin")
y2 = wave(x, "cos")

plt.figure(figsize=(9, 4))
plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")
plt.xlabel("x")
plt.ylabel("value")
plt.title("Two Waves")
plt.legend()
plt.grid(True)
plt.show()
```

<div class="sv-section"><div class="sv-section-title">🧮 16. Build a Simple Calculator</div><div class="sv-section-sub">A complete beginner project using loops, input, conditions, and validation.</div></div>

```python
# Interactive project — stop by choosing 0
while True:
    print("\n--- SIMPLE CALCULATOR ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("0. Exit")

    choice = input("Choose: ")

    if choice == "0":
        print("Goodbye 👋")
        break

    if choice not in ["1", "2", "3", "4"]:
        print("Invalid option. Try again.")
        continue

    num1 = float(input("First number: "))
    num2 = float(input("Second number: "))

    if choice == "1":
        print("Result:", num1 + num2)
    elif choice == "2":
        print("Result:", num1 - num2)
    elif choice == "3":
        print("Result:", num1 * num2)
    elif choice == "4":
        if num2 == 0:
            print("Cannot divide by zero!")
        else:
            print("Result:", num1 / num2)
```

```python
# Interactive game
import random
choices = ["rock", "paper", "scissors"]
computer = random.choice(choices)
player = input("rock, paper or scissors? ").lower()
print("Computer:", computer)
if player == computer:
    print("😐 Draw!")
elif ((player == "rock" and computer == "scissors") or
      (player == "paper" and computer == "rock") or
      (player == "scissors" and computer == "paper")):
    print("😎 You win!")
else:
    print("🤖 Computer wins!")
```

### 🛸 Final Space Calculator

```python
print("🛸 SPACE CALCULATOR 🛸")
number1 = float(input("First number: "))
operator = input("+  -  *  / : ")
number2 = float(input("Second number: "))

if operator == "+":
    answer = number1 + number2
elif operator == "-":
    answer = number1 - number2
elif operator == "*":
    answer = number1 * number2
elif operator == "/":
    answer = "Cannot divide by zero" if number2 == 0 else number1 / number2
else:
    answer = "Unknown command 🤖"

print("Answer =", answer)
```
